# Experiment 01: Document Parsing & Layout Recognition

This experiment compares **Tesseract OCR**, **PaddleOCR**, and **Google Gemini API** for parsing questionnaire layouts and detecting form field structures (checkboxes, radio buttons, text areas).

## Objectives
1. Benchmark bounding box IoU and field type extraction accuracy.
2. Track processing latency and resource utilization using **MLflow**.
3. Compare open-source local models (PaddleOCR/Tesseract) vs API-based multimodal LLMs (Gemini).

```python
import os
import time
import json
import cv2
import numpy as np
import mlflow

# Set MLflow experiment name
mlflow.set_experiment("Document_Parsing_Comparison")
print("MLflow experiment initialized.")
```

## 1. Tesseract OCR Baseline Evaluation

```python
import pytesseract

with mlflow.start_run(run_name="Tesseract_Baseline"):
    mlflow.log_param("engine", "tesseract")
    mlflow.log_param("psm_mode", 6)
    
    start_time = time.time()
    # Dummy test image processing simulation
    test_img = np.zeros((500, 500, 3), dtype=np.uint8)
    text = pytesseract.image_to_string(test_img)
    latency = time.time() - start_time
    
    mlflow.log_metric("latency_seconds", latency)
    mlflow.log_metric("detected_fields_count", 0)
    print(f"Tesseract Run Complete. Latency: {latency:.4f}s")
```

## 2. PaddleOCR Layout & Field Detection

```python
# PaddleOCR Experiment Run
with mlflow.start_run(run_name="PaddleOCR_Layout_Parsing"):
    mlflow.log_param("engine", "paddleocr")
    mlflow.log_param("lang", "en")
    mlflow.log_param("use_angle_cls", True)
    
    start_time = time.time()
    # Simulated detection metrics for layout parsing
    simulated_iou = 0.88
    simulated_accuracy = 0.92
    latency = time.time() - start_time
    
    mlflow.log_metric("bounding_box_iou", simulated_iou)
    mlflow.log_metric("field_classification_acc", simulated_accuracy)
    mlflow.log_metric("latency_seconds", latency)
    print(f"PaddleOCR Run Complete. Field Accuracy: {simulated_accuracy}")
```

## 3. Gemini API Zero-Shot Questionnaire Schema Extraction

```python
# Gemini API Multimodal Experiment Run
with mlflow.start_run(run_name="Gemini_1.5_Flash_Multimodal"):
    mlflow.log_param("engine", "gemini-1.5-flash")
    mlflow.log_param("prompt_strategy", "zero_shot_json_schema")
    
    start_time = time.time()
    # Simulated Gemini extraction metrics
    simulated_accuracy = 0.96
    simulated_cost_per_doc = 0.0015
    latency = time.time() - start_time
    
    mlflow.log_metric("field_classification_acc", simulated_accuracy)
    mlflow.log_metric("estimated_cost_usd", simulated_cost_per_doc)
    mlflow.log_metric("latency_seconds", latency)
    print(f"Gemini API Run Complete. Field Accuracy: {simulated_accuracy}")
```
